```python
def knn(X, Y, queryPoint, k):
    distances = np.sqrt(np.sum((queryPoint-X)**2, axis=1))
    distances = [(distances[i], Y[i]) for i in range(len(distances))]
    distances = sorted(distances)
    distances = distances[:k]
    distances = np.array(distances)
    classes_counts = np.unique(distances[:,1], return_counts=True)
    index = classes_counts[1].argmax()
    pred = classes_counts[0][index]
    return int(pred),distances
```

```python
pred, neighbors = knn(X_train_scaled, y_sm, X_test_scaled[50], 5)

print(f' k nearest neighbors: {neighbors}')
print(f' Predicted class: {pred}')
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=5, metric='euclidean')
```

```python
knn.fit(X_train_scaled, y_sm)
```

```python
knn.score(X_test_scaled, y_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
y_pred = knn.predict(X_test_scaled)
print(classification_report(y_test, y_pred))
```
