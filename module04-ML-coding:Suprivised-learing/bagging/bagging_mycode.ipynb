```python
#function for row sampling

def sample_rows(df, percent):
    return df.sample(int(percent * df.shape[0]), replace=True)
```

```python
import random
random.sample(df.columns.tolist()[:-1], int(0.5*(df.shape[1]-1)))
```

```python
#function fetaure sampling

def sample_features(df, percent):
    cols= random.sample(df.columns.tolist()[:-1], int(percent*(df.shape[1]-1)))
    new_df = df[cols]
    new_df['target'] = df['target']
    return cols
```

```python
sample_features(df, 0.2)
```

```python
def combined_samples(df, row_percent, col_pecent):
    new_df = sample_rows(df, row_percent)
    return sample_features(new_df, col_pecent)
```

```python
combined_samples(df, 0.5, 0.5)
```
