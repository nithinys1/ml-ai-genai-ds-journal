```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

!curl -L "https://drive.google.com/uc?export=download&id=1UpLnYA48Vy_lGUMMLG-uQE1gf_Je12Lh" -o univariate_data_price_model.csv
```

```python
df = pd.read_csv("univariate_data_price_model.csv")
df.head()
```

```python
X = df["model"].values
Y = df["selling_price"].values
Y.shape
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model 

```

```python
type(model)
```

```python
X = X.reshape(X.size, 1 )
Y = Y.reshape(Y.size, 1 )
X.shape, Y.shape
```

```python
model.fit(X, Y)
```

```python
y_hat = model.predict(X)
y_hat
```

```python
model.coef_, model.intercept_
```

```python
fig = plt.figure()
plt.scatter(X, Y, label="Data")
plt.scatter(X, y_hat, color="orange", label="Predicted line")
plt.legend()
```
