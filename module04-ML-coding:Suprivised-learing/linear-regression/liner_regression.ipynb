```python
df = pd.read_csv('train-cars24-car-price.csv')
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.title('MaxPower vs price')
sns.scatterplot(x='max_power', y='selling_price', data=df)
```

```python
def estimate_Charge(max_power, m, c):
    return m * max_power + c
```

```python
m = 50
c = 100

df.max_power
```

```python
maxpower2 = df.max_power
estimated_price = estimate_Charge(maxpower2, m, c)
estimated_price
```

```python
plt.figure(figsize=(10, 6))
plt.plot(maxpower2, estimated_price, color='red', label='Estimated Price')
```

```python
target = df.selling_price
plt.figure(figsize=(10, 6))
plt.plot(maxpower2 , estimated_price, color='red', label='Estimated Price')
plt.scatter(maxpower2, target, color='blue', label='Actual Price')
plt.xlabel('Max Power')
plt.ylabel('Price')
plt.title('Linear Regression: Max Power vs Price')
plt.legend()
plt.show()
```

```python
def try_parmeters(m, c):
    maxpower2 = df.max_power
    target = df.selling_price
    estimated_price = estimate_Charge(maxpower2, m, c)
    plt.figure(figsize=(10, 6))
    plt.plot(maxpower2 , estimated_price, color='red', label='Estimated Price')
    plt.scatter(maxpower2, target, color='blue', label='Actual Price')
    plt.xlabel('Max Power')
    plt.ylabel('Price')
    plt.title(f'Linear Regression: Max Power vs Price (m={m}, c={c})')
    plt.legend()
   
```

```python
try_parmeters(0.5 , 1)
```
