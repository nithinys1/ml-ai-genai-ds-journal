# Q6: Measure the Distance - Euclidean & Manhattan Distance

Calculate Euclidean and Manhattan distances between a query point and multiple datapoints.


```python
import numpy as np

# Define the query point and datapoints
query_point = [5, 10, 15]
datapoints = [[4, 13, 2], [9, 8, 11], [14, 4, 2], [4, 12, 1], [2, 8, 3]]

# Convert to numpy arrays for easier computation
query = np.array(query_point)
data = np.array(datapoints)

print(f"Query Point: {query_point}")
print(f"Datapoints: {datapoints}")
```

```python
# Calculate Euclidean Distance
# Formula: √(Σ(xi - yi)²)

euclidean_distances = []
for datapoint in data:
    distance = np.sqrt(np.sum((query - datapoint) ** 2))
    euclidean_distances.append(round(distance, 2))

print("Euclidean Distances:")
print(euclidean_distances)
```

```python
# Calculate Manhattan Distance
# Formula: Σ|xi - yi|

manhattan_distances = []
for datapoint in data:
    distance = np.sum(np.abs(query - datapoint))
    manhattan_distances.append(round(distance, 2))

print("\nManhattan Distances:")
print(manhattan_distances)
```

```python
# Reusable function for calculating distances

def calculate_distances(query_point, datapoints):
    """
    Calculate Euclidean and Manhattan distances between a query point and multiple datapoints.
    
    Args:
        query_point: A list/array representing the query point
        datapoints: A list of lists/arrays representing the datapoints
        
    Returns:
        euclidean_distances: List of Euclidean distances rounded to 2 decimal places
        manhattan_distances: List of Manhattan distances rounded to 2 decimal places
    """
    query = np.array(query_point)
    data = np.array(datapoints)
    
    # Euclidean distances
    euclidean = [round(np.sqrt(np.sum((query - point) ** 2)), 2) for point in data]
    
    # Manhattan distances
    manhattan = [round(np.sum(np.abs(query - point)), 2) for point in data]
    
    return euclidean, manhattan

# Test the function
euc_dist, man_dist = calculate_distances(query_point, datapoints)
print("Function Results:")
print(f"Euclidean Distances: {euc_dist}")
print(f"Manhattan Distances: {man_dist}")
```

```python
# Distance calculation functions with numpy

import numpy as np

def euclideanDistance(point, X):
    """
    Calculate Euclidean distance between a point and all rows in X.
    
    Args:
        point: Query point (numpy array or list)
        X: Array of datapoints (2D numpy array)
        
    Returns:
        List of Euclidean distances rounded to 2 decimal places
    """
    point = np.asarray(point)
    X = np.asarray(X)
    
    distances = []
    for x in X:
        distance = np.sqrt(np.sum((point - x) ** 2))
        distances.append(round(distance, 2))
    
    return distances


def manhattanDistance(point, X):
    """
    Calculate Manhattan distance between a point and all rows in X.
    
    Args:
        point: Query point (numpy array or list)
        X: Array of datapoints (2D numpy array)
        
    Returns:
        List of Manhattan distances rounded to 2 decimal places
    """
    point = np.asarray(point)
    X = np.asarray(X)
    
    distances = []
    for x in X:
        distance = np.sum(np.abs(point - x))
        distances.append(round(distance, 2))
    
    return distances


# Test the functions
query_point_test = [5, 10, 15]
datapoints_test = [[4, 13, 2], [9, 8, 11], [14, 4, 2], [4, 12, 1], [2, 8, 3]]

euc_result = euclideanDistance(query_point_test, datapoints_test)
man_result = manhattanDistance(query_point_test, datapoints_test)

print("Euclidean Distances:", euc_result)
print("Manhattan Distances:", man_result)
```
