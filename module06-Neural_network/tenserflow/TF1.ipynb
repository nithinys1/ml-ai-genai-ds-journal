```python
# Complete, minimal NN with manual forward + backward (TensorFlow 2.x)
import tensorflow as tf
import numpy as np
import time

# Reproducibility
tf.random.set_seed(42)
np.random.seed(42)

# -----------------------
# 1) Create synthetic dataset (simple linear function + noise)
# -----------------------
N = 500            # number of samples
X = np.random.uniform(-1, 1, size=(N, 1)).astype(np.float32)   # shape (N, 1)
true_w = 3.0
true_b = 2.0
noise = np.random.normal(0.0, 0.1, size=(N, 1)).astype(np.float32)
Y = (true_w * X + true_b + noise).astype(np.float32)          # shape (N, 1)

# Convert to tf tensors
X_tf = tf.constant(X)   # input
Y_tf = tf.constant(Y)   # targets

# -----------------------
# 2) Define network architecture and initialize weights (as tf.Variable)
#    Simple 1-hidden-layer network: input(1) -> hidden(8) -> output(1)
# -----------------------
input_dim = 1
hidden_units = 8
output_dim = 1

# Initialize weights and biases
W1 = tf.Variable(tf.random.normal([input_dim, hidden_units], stddev=0.5), name="W1")
b1 = tf.Variable(tf.zeros([hidden_units]), name="b1")

W2 = tf.Variable(tf.random.normal([hidden_units, output_dim], stddev=0.5), name="W2")
b2 = tf.Variable(tf.zeros([output_dim]), name="b2")

# -----------------------
# 3) Define forward pass (pure tensor ops)
# -----------------------
def forward_pass(x):
    """Compute network output for input x."""
    # Linear -> ReLU -> Linear (no activation on output for regression)
    z1 = tf.matmul(x, W1) + b1           # shape (batch, hidden_units)
    a1 = tf.nn.relu(z1)                  # activation
    out = tf.matmul(a1, W2) + b2         # shape (batch, output_dim)
    return out

# -----------------------
# 4) Define loss function (Mean Squared Error)
# -----------------------
def mse_loss(y_pred, y_true):
    return tf.reduce_mean(tf.square(y_pred - y_true))

# -----------------------
# 5) Training hyperparameters
# -----------------------
learning_rate = 0.1
epochs = 200
batch_size = 32
steps_per_epoch = int(np.ceil(N / batch_size))

# -----------------------
# 6) Manual training loop with GradientTape (explicit backward pass)
# -----------------------
start_time = time.time()
for epoch in range(1, epochs + 1):
    # Shuffle the data indices
    idx = np.random.permutation(N)
    X_shuffled = X_tf.numpy()[idx]
    Y_shuffled = Y_tf.numpy()[idx]

    epoch_loss = 0.0

    for step in range(steps_per_epoch):
        start = step * batch_size
        end = min(start + batch_size, N)
        x_batch = tf.constant(X_shuffled[start:end])
        y_batch = tf.constant(Y_shuffled[start:end])

        # Record operations for automatic differentiation
        with tf.GradientTape() as tape:
            y_pred = forward_pass(x_batch)          # forward
            loss = mse_loss(y_pred, y_batch)        # compute loss

        # Backward: compute gradients of loss w.r.t. all trainable variables
        grads = tape.gradient(loss, [W1, b1, W2, b2])

        # Manual SGD parameter update: param = param - lr * grad
        W1.assign_sub(learning_rate * grads[0])
        b1.assign_sub(learning_rate * grads[1])
        W2.assign_sub(learning_rate * grads[2])
        b2.assign_sub(learning_rate * grads[3])

        epoch_loss += loss.numpy() * (end - start)

    # average loss over dataset
    epoch_loss /= N

    # print some progress
    if epoch % 20 == 0 or epoch == 1:
        print(f"Epoch {epoch:3d}/{epochs} - Loss: {epoch_loss:.6f}")

end_time = time.time()
print(f"Training finished in {end_time - start_time:.3f} seconds")

# -----------------------
# 7) Evaluate learned parameters and final loss on full dataset
# -----------------------
y_pred_final = forward_pass(X_tf)
final_loss = mse_loss(y_pred_final, Y_tf).numpy()
print("\nFinal MSE on training set:", final_loss)

# Print learned linear approximation for sanity (since target is linear)
# Note: The small hidden net should be able to represent a linear mapping approx.
# We can run a simple least-square on predictions to see approximate slope/intercept.
coef = np.polyfit(X.flatten(), y_pred_final.numpy().flatten(), 1)
print(f"Learned linear fit (approx): y = {coef[0]:.3f}x + {coef[1]:.3f}")
print(f"True function was y = {true_w}x + {true_b}")

# -----------------------
# 8) Inference example: predict on a new point
# -----------------------
x_test = tf.constant([[0.5]], dtype=tf.float32)
y_test_pred = forward_pass(x_test).numpy().squeeze()
y_test_true = true_w * 0.5 + true_b
print(f"\nFor x = 0.5, predicted y = {y_test_pred:.3f}, true y = {y_test_true:.3f}")
```
