```python
import numpy as np 

class SimpleNeuron:
    def __init__(self, n_inputs=2, weights=None, bias=None):
        if weights is None:
            self.weights = np.random.rand(n_inputs) * 0.01
        else: 
            self.weights = np.array(weights)
        if bias is None:
            self.bias = 0.0
        else:
            self.bias = bias
        print(f" Neuron intialized")
        print(f" Weights: {self.weights}")
        print(f" Bias: {self.bias}")
        
    def sigmoid (self, z):
        return 1 / (1 + np.exp(-z))
    
    def Process(self, inputs):
        z = np.dot(inputs, self.weights) + self.bias
        output = self.sigmoid(z)
        return output, z
    
    def decide (self, inputs, threshold = 0.5):
        probability, z = self.Process(inputs)
        decision = probability >= threshold
        return decision, probability, z
    
    
```

```python
class TouchDecisionNeuron(SimpleNeuron):
    def __init__(self):
        super().__init__(n_inputs=2, weights=[-2.0, 1.5], bias=-0.5)
        
        print(f" TouchDecisionNeuron initialized with weights: {self.weights} and bias: {self.bias}")
        
    def should_i_topuch(self, tempreature, Familiarity):
        inputs = np.array([tempreature, Familiarity])
        decision, probability, z = self.decide(inputs)
        if decision:
            return f"Touch the object! (Probability: {probability:.2f}, z: {z:.2f})"
        else:
            return f"Do not touch the object! (Probability: {probability:.2f}, z: {z:.2f})"
        
touch_neuron = TouchDecisionNeuron()

print("\n" + "-"*40)
print(touch_neuron.should_i_topuch(0.8, 0.3))
```

```python
# ============================================
# EXAMPLE 2: Generic Binary Classification
# ============================================

spam_neuron = SimpleNeuron(n_inputs=3, weights=[1.0, 1.0, 1.0], bias=-1.5)
print(f"SpamNeuron initialized with weights: {spam_neuron.weights} and bias: {spam_neuron.bias}")

test_emails = [
    [1, 0, 1],  # Spam
    [0, 1, 0],  # Not Spam
    [1, 1, 1],  # Spam
]

print("\n" + "-"*40)
for features in test_emails:
    _, probability, z = spam_neuron.decide(features)
    is_spam = "Spam" if probability >= 0.5 else "Not Spam"
    print(f"Email features: {features} => Decision: {is_spam} (Probability: {probability:.2f}, z: {z:.2f})")
```
