```python
def init_params(layer_sizes):

  np.random.seed(3)
  params = {}
  L = len(layer_sizes)

  for l in range(1, L):

    params['W' + str(l)] = np.ones((layer_sizes[l-1], layer_sizes[l]))*0.1
    params['b' + str(l)] = np.zeros((layer_sizes[l], 1))


  return params
```

```python
init_params([2,2,1])
```

```python
def linear_forward_step(activ_prev, weights, bias):

  Z = np.dot(weights.T, activ_prev) + bias

  return Z
```

```python
# Forward Prop
def forward_pass(input_data, params):

  A = input_data
  L = len(params) // 2                  # number of layers in the neural network

  for l in range(1, L+1):
    activ_prev = A
    Wl = params['W' + str(l)]
    bl = params['b' + str(l)]
    #print("A"+str(l-1)+": ", A_prev)
    #print("W"+str(l)+": ", Wl)
    #print("b"+str(l)+": ", bl)
    #print("--"*20)

    A = linear_forward_step(activ_prev, Wl, bl)
    #print("A"+str(l)+": ", A)
    #print("**"*20)

  return A,activ_prev
```

```python
input_data = df[['cgpa', 'profile_score']].values[0].reshape(2,1) # Shape(no of features, no. of training example)
target = df[['lpa']].values[0][0]

# Parameter initialization
params = init_params([2,2,1])

y_pred,hidden_activ = forward_pass(input_data, params)
```

```python
def update_params(params,target,y_pred,hidden_activ,input_data):
  params['W2'][0][0] = params['W2'][0][0] + (0.001 * 2 * (target - y_pred)*hidden_activ[0][0])
  params['W2'][1][0] = params['W2'][1][0] + (0.001 * 2 * (target - y_pred)*hidden_activ[1][0])
  params['b2'][0][0] = params['W2'][1][0] + (0.001 * 2 * (target - y_pred))

  params['W1'][0][0] = params['W1'][0][0] + (0.001 * 2 * (target - y_pred)*params['W2'][0][0]*input_data[0][0])
  params['W1'][0][1] = params['W1'][0][1] + (0.001 * 2 * (target - y_pred)*params['W2'][0][0]*input_data[1][0])
  params['b1'][0][0] = params['b1'][0][0] + (0.001 * 2 * (target - y_pred)*params['W2'][0][0])

  params['W1'][1][0] = params['W1'][1][0] + (0.001 * 2 * (target - y_pred)*params['W2'][1][0]*input_data[0][0])
  params['W1'][1][1] = params['W1'][1][1] + (0.001 * 2 * (target - y_pred)*params['W2'][1][0]*input_data[1][0])
  params['b1'][1][0] = params['b1'][1][0] + (0.001 * 2 * (target - y_pred)*params['W2'][1][0])
```

```python
input_data = df[['cgpa', 'profile_score']].values[0].reshape(2,1) # Shape(no of features, no. of training example)
target = df[['lpa']].values[0][0]

# Parameter initialization
params = init_params([2,2,1])

y_pred,hidden_activ = forward_pass(input_data,params)
y_pred = y_pred[0][0]

update_params(params,target,y_pred,hidden_activ,input_data)

params
```

```python
# epochs implementation

params = init_params([2,2,1])
epochs = 5

for i in range(epochs):

  Loss = []

  for j in range(df.shape[0]):

    input_data = df[['cgpa', 'profile_score']].values[j].reshape(2,1) # Shape(no of features, no. of training example)
    target = df[['lpa']].values[j][0]

    # Parameter initialization


    y_pred,hidden_activ = forward_pass(input_data,params)
    y_pred = y_pred[0][0]

    update_params(params,target,y_pred,hidden_activ,input_data)

    Loss.append((target-y_pred)**2)

  print('Epoch - ',i+1,'Loss - ',np.array(Loss).mean())

params
```
