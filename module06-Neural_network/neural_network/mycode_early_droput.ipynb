```python
import tensorflow as tf
import numpy as np
import pandas as pd
from pylab import rcParams
import matplotlib.pyplot as plt
import warnings
from mlxtend.plotting import plot_decision_regions
from matplotlib.colors import ListedColormap
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dropout
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_circles
import seaborn as sns
```

```python
X, y = make_circles(n_samples=100, noise=0.1, random_state=1)
```

```python
sns.scatterplot(x=X[:,0],y=X[:,1],hue=y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=2)
```

```python
model = Sequential()

model.add(Dense(256, input_dim=2, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
```

```python
model.compile(loss='binary_crossentropy', optimizer='sgd', metrics=['accuracy'])
```

```python
history = model.fit(X_train, y_train, validation_data=(X_test, y_test), epochs=3500, verbose=0)
```

```python
plt.plot(history.history['loss'], label='train')
plt.plot(history.history['val_loss'], label='test')
plt.legend()
plt.show()
```

```python
model = Sequential()

model.add(Dense(256, input_dim=2, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
```

```python
model.compile(loss='binary_crossentropy', optimizer='sgd', metrics=['accuracy'])
```

```python
callback = EarlyStopping(
    monitor="accuracy",
    patience=10,
    restore_best_weights=True
)
```

```python
history = model.fit(X_train, y_train, validation_data=(X_test, y_test), epochs=3500, callbacks=callback)
```

```python
print("Stopped at epoch:", callback.stopped_epoch)
```

```python
print("Best val_loss:", callback.best)
```

```python
model.evaluate(X_test, y_test)
```

```python
radius = np.sqrt((X ** 2).sum(axis=1))
y_dataset = (radius > 0.35).astype("float32")

X_train_dataset, X_test_dataset, y_train_dataset, y_test_dataset = train_test_split(
    X,
    y_dataset,
    test_size=0.2,
    random_state=2,
    stratify=y_dataset
)

model_dropout = Sequential([
    tf.keras.Input(shape=(2,)),
    Dense(256, activation="relu"),
    Dropout(0.5),
    Dense(128, activation="relu"),
    Dropout(0.5),
    Dense(1, activation="sigmoid")
])

model_dropout.compile(
    loss="binary_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

history_dropout = model_dropout.fit(
    X_train_dataset,
    y_train_dataset,
    validation_data=(X_test_dataset, y_test_dataset),
    epochs=100,
    batch_size=16,
    verbose=0
)

test_loss, test_accuracy = model_dropout.evaluate(
    X_test_dataset,
    y_test_dataset,
    verbose=0
)
print(f"Test loss: {test_loss:.4f}")
print(f"Test accuracy: {test_accuracy:.4f}")

plt.scatter(X[:, 0], X[:, 1], c=y_dataset, cmap="coolwarm")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Pasted dataset with radius-based labels")
plt.show()
```

```python
model_no_dropout = Sequential([
    tf.keras.Input(shape=(2,)),
    Dense(256, activation="relu"),
    Dense(128, activation="relu"),
    Dense(1, activation="sigmoid")
])

model_no_dropout.compile(
    loss="binary_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

history_no_dropout = model_no_dropout.fit(
    X_train_dataset,
    y_train_dataset,
    validation_data=(X_test_dataset, y_test_dataset),
    epochs=100,
    batch_size=16,
    verbose=0
)

_, accuracy_no_dropout = model_no_dropout.evaluate(
    X_test_dataset,
    y_test_dataset,
    verbose=0
)

_, accuracy_dropout = model_dropout.evaluate(
    X_test_dataset,
    y_test_dataset,
    verbose=0
)

x_min, x_max = X[:, 0].min() - 0.1, X[:, 0].max() + 0.1
y_min, y_max = X[:, 1].min() - 0.1, X[:, 1].max() + 0.1
xx, yy = np.meshgrid(
    np.linspace(x_min, x_max, 250),
    np.linspace(y_min, y_max, 250)
)
grid = np.c_[xx.ravel(), yy.ravel()]

prediction_no_dropout = model_no_dropout.predict(grid, verbose=0).reshape(xx.shape)
prediction_dropout = model_dropout.predict(grid, verbose=0).reshape(xx.shape)

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

axes[0].scatter(X[:, 0], X[:, 1], c=y_dataset, cmap="coolwarm", edgecolors="black", s=35)
axes[0].set_title("Original pasted data")

axes[1].contourf(xx, yy, prediction_no_dropout, levels=[0, 0.5, 1], alpha=0.35, cmap="coolwarm")
axes[1].scatter(X[:, 0], X[:, 1], c=y_dataset, cmap="coolwarm", edgecolors="black", s=35)
axes[1].set_title(f"Without dropout\nAccuracy: {accuracy_no_dropout:.3f}")

axes[2].contourf(xx, yy, prediction_dropout, levels=[0, 0.5, 1], alpha=0.35, cmap="coolwarm")
axes[2].scatter(X[:, 0], X[:, 1], c=y_dataset, cmap="coolwarm", edgecolors="black", s=35)
axes[2].set_title(f"With dropout\nAccuracy: {accuracy_dropout:.3f}")

for axis in axes:
    axis.set_xlabel("Feature 1")
    axis.set_ylabel("Feature 2")

plt.tight_layout()
plt.show()
```
