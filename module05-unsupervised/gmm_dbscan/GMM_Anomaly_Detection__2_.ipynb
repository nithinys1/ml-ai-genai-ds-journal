```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score

# 1. Generate synthetic dataset
X, y_true = make_blobs(n_samples=500, centers=4, cluster_std=0.7, random_state=42)
```

```python
plt.scatter(X[:, 0], X[:, 1], c=y_true, cmap='viridis')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Synthetic Dataset')
plt.show()
```

```python
X
```

```python
# 2. Try different numbers of clusters (K)
sil_scores = []
K_range = range(2, 8)  # test from 2 to 7 clusters

for k in K_range:
    gmm = GaussianMixture(n_components=k, random_state=42)
    labels = gmm.fit_predict(X)
    score = silhouette_score(X, labels)
    sil_scores.append(score)
    print(f"K={k}, Silhouette Score={score:.3f}")

# 3. Best K
best_k = K_range[np.argmax(sil_scores)]
print("\n✅ Best number of clusters =", best_k)

# 4. Plot silhouette scores
plt.plot(K_range, sil_scores, marker='o')
plt.xlabel("Number of clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Scores for different K in GMM")
plt.show()

```

```python
# 5. Fit GMM with best K and visualize
gmm = GaussianMixture(n_components=best_k, random_state=42)
labels = gmm.fit_predict(X)

plt.scatter(X[:, 0], X[:, 1], c=labels, cmap="viridis", s=30, alpha=0.6)
plt.scatter(gmm.means_[:, 0], gmm.means_[:, 1], marker='x', s=200, c="red", label="Cluster centers")
plt.title(f"GMM Clustering with K={best_k}")
plt.legend()
plt.show()
```
